## Initial plan

* Dataset - [BDNeuro-MRI](https://data.mendeley.com/datasets/zwr4ntf94j/7)
* Goal - predict tumor classes (Glioma Tumor, Meningioma Tumor, Pituitary Tumor, No Tumor)
    * Accuracy per class
    * *Weighted loss to combat FN?*
* Model architecture - Deep CNN, softmax output
* Training setup - set seed (1738) for reproducibility, 70/15/15 train/val/test split (given in data), cross entropy loss?, early stopping with checkpoints, data augmentation
* ROC curve

### Steps

**1) Data exploration** - explore data, check for imbalances or other things that affect model/eval choices

**2) Eval setup** - decide metrics

**3) Preprocessing and dataloader** - preprocess data and build dataloader

**4) Initial model** - build small model with semi-guessed parameters to show proof of concept

**5) Input experimentation** - compare different input setups
* grayscale vs RGB
* resolution

**6) Optimization tuning**
* learning rate
* optimizer
* batch size
* schedule?
* loss function?

**7) Hyperparameter tuning**
* kernel sizes
* downsampling?
* depth
    * batchnorm?
    * skip connections?
* width

**8) Regularisation** - help model generalize if val error is significantly higher than train error
* augmentation?
* weight decay?
* dropout

**9) LR again** - see if learning rate should be adjusted after other tuning is done

**10) ROC curve** - pick cutoff for ROC curve

**11) Test model** - run final model on test set

**12) Error analysis** - try to see how it fails when it does
